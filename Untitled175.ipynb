```python
import os
import json
import yaml
import pickle
import warnings
from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import joblib

from sklearn.model_selection import train_test_split, KFold, cross_val_predict
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.dummy import DummyRegressor
from sklearn.ensemble import (
    RandomForestRegressor,
    GradientBoostingRegressor
)
from sklearn.tree import DecisionTreeRegressor
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout, Input
from tensorflow.keras.callbacks import EarlyStopping
from tensorflow.keras.optimizers import Adam

warnings.filterwarnings("ignore", category=FutureWarning)

# ================================================================
# 1. PROJECT CONFIGURATION
# ================================================================

ROOT = Path(
    r"C:\Users\sagni\Downloads\AI-Based Coal Mine Production Analysis, Mining Company Performance Assessment, and Production Optimization System"
)

CSV_PATH = ROOT / "rs_session-241_au482_1.1.csv"

ROOT.mkdir(parents=True, exist_ok=True)

SEED = 42
np.random.seed(SEED)
tf.random.set_seed(SEED)

if not CSV_PATH.exists():
    raise FileNotFoundError(
        f"Dataset not found: {CSV_PATH}\n"
        "Check the folder and CSV filename."
    )

print("Project Directory:", ROOT)
print("Dataset:", CSV_PATH)

# ================================================================
# 2. LOAD DATASET
# ================================================================

try:
    df = pd.read_csv(CSV_PATH, encoding="utf-8-sig")
except UnicodeDecodeError:
    df = pd.read_csv(CSV_PATH, encoding="latin1")

df.columns = (
    df.columns.astype(str)
    .str.strip()
    .str.replace(r"\s+", " ", regex=True)
)

df = df.dropna(how="all").copy()

print("\nDataset Shape:", df.shape)
print("\nAvailable Columns:")
print(df.columns.tolist())
print("\nFirst Five Rows:")
print(df.head())

# ================================================================
# 3. AUTOMATIC COLUMN IDENTIFICATION
# ================================================================

def find_column(columns, keywords):
    for keyword in keywords:
        matches = [
            col for col in columns
            if keyword.lower() in col.lower()
        ]
        if matches:
            return matches[0]
    return None

mine_col = find_column(
    df.columns,
    ["Name of the Mine", "Mine Name", "Mine"]
)

company_col = find_column(
    df.columns,
    ["Type of WCL", "Company", "Coalfield", "Organisation"]
)

target_col = find_column(
    df.columns,
    ["Quantity for Coal production", "Coal Production", "Quantity"]
)

if not all([mine_col, company_col, target_col]):
    raise ValueError(
        "Could not identify required columns.\n"
        f"Columns found: {df.columns.tolist()}\n"
        f"Mine: {mine_col}\n"
        f"Company: {company_col}\n"
        f"Target: {target_col}"
    )

print("\nIdentified Columns:")
print("Mine:", mine_col)
print("Company:", company_col)
print("Production:", target_col)

# ================================================================
# 4. DATA CLEANING
# ================================================================

data = df[[mine_col, company_col, target_col]].copy()

data.columns = [
    "Mine",
    "Company",
    "Coal_Production"
]

for col in ["Mine", "Company"]:
    data[col] = (
        data[col]
        .astype("string")
        .str.strip()
        .replace("", pd.NA)
        .fillna("Unknown")
        .astype(str)
    )

data["Coal_Production"] = pd.to_numeric(
    data["Coal_Production"]
    .astype(str)
    .str.replace(",", "", regex=False)
    .str.strip(),
    errors="coerce"
)

initial_rows = len(data)

data = data.dropna(subset=["Coal_Production"])
data = data[data["Coal_Production"] >= 0]
data = data.drop_duplicates().reset_index(drop=True)

print("\nRows Removed:", initial_rows - len(data))
print("Clean Dataset Shape:", data.shape)

if len(data) < 15:
    raise ValueError(
        "Insufficient valid records for model evaluation."
    )

data.to_csv(
    ROOT / "cleaned_coal_production.csv",
    index=False
)

# ================================================================
# 5. EXPLORATORY DATA ANALYSIS
# ================================================================

company_summary = (
    data.groupby("Company")["Coal_Production"]
    .agg(["count", "sum", "mean", "median", "std"])
    .reset_index()
    .rename(columns={
        "count": "Mine_Count",
        "sum": "Total_Production",
        "mean": "Average_Production",
        "median": "Median_Production",
        "std": "Production_Std"
    })
    .sort_values("Total_Production", ascending=False)
)

company_summary.to_csv(
    ROOT / "company_performance.csv",
    index=False
)

print("\nCompany Performance:")
print(company_summary)

# ================================================================
# 6. VISUALIZATIONS
# ================================================================

sns.set_theme(style="whitegrid")

plt.figure(figsize=(12, 6))
sns.barplot(
    data=company_summary,
    x="Company",
    y="Total_Production",
    color="steelblue"
)
plt.xticks(rotation=45, ha="right")
plt.title("Company-Wise Coal Production")
plt.tight_layout()
plt.savefig(
    ROOT / "company_production_graph.png",
    dpi=300
)
plt.show()

plt.figure(figsize=(10, 6))
sns.histplot(
    data["Coal_Production"],
    bins=25,
    kde=True
)
plt.title("Coal Production Distribution")
plt.xlabel("Coal Production")
plt.tight_layout()
plt.savefig(
    ROOT / "production_distribution.png",
    dpi=300
)
plt.show()

# ================================================================
# 7. FEATURE ENGINEERING
# ================================================================

# Mine names are identifiers and can cause memorization.
# We therefore use company as the predictive feature.
# The mine name is retained for reporting.

X = data[["Company"]].copy()
y = data["Coal_Production"].astype(float)

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=SEED
)

def make_preprocessor():
    categorical_pipeline = Pipeline([
        (
            "imputer",
            SimpleImputer(strategy="most_frequent")
        ),
        (
            "encoder",
            OneHotEncoder(
                handle_unknown="ignore",
                sparse_output=False
            )
        )
    ])

    return ColumnTransformer([
        (
            "categorical",
            categorical_pipeline,
            ["Company"]
        )
    ])

# ================================================================
# 8. TRAIN MACHINE LEARNING MODELS
# ================================================================

models = {
    "Baseline": DummyRegressor(
        strategy="mean"
    ),
    "Decision_Tree": DecisionTreeRegressor(
        max_depth=5,
        min_samples_leaf=3,
        random_state=SEED
    ),
    "Random_Forest": RandomForestRegressor(
        n_estimators=200,
        max_depth=8,
        min_samples_leaf=2,
        random_state=SEED,
        n_jobs=-1
    ),
    "Gradient_Boosting": GradientBoostingRegressor(
        n_estimators=150,
        learning_rate=0.05,
        max_depth=3,
        random_state=SEED
    )
}

trained_models = {}
evaluation_results = []

def calculate_metrics(actual, predicted):
    return {
        "MAE": float(mean_absolute_error(actual, predicted)),
        "RMSE": float(
            np.sqrt(mean_squared_error(actual, predicted))
        ),
        "R2": float(r2_score(actual, predicted))
    }

for name, estimator in models.items():

    pipeline = Pipeline([
        ("preprocessor", make_preprocessor()),
        ("regressor", estimator)
    ])

    pipeline.fit(X_train, y_train)

    predictions = pipeline.predict(X_test)

    metrics = calculate_metrics(
        y_test,
        predictions
    )

    evaluation_results.append({
        "Model": name,
        **metrics
    })

    trained_models[name] = pipeline

    joblib.dump(
        pipeline,
        ROOT / f"{name.lower()}_model.pkl"
    )

    print(f"\n{name}")
    print("MAE:", round(metrics["MAE"], 4))
    print("RMSE:", round(metrics["RMSE"], 4))
    print("R2:", round(metrics["R2"], 4))

# ================================================================
# 9. SELECT BEST MACHINE LEARNING MODEL
# ================================================================

ml_results = pd.DataFrame(evaluation_results)

ml_results = ml_results.sort_values(
    "RMSE",
    ascending=True
).reset_index(drop=True)

best_model_name = ml_results.iloc[0]["Model"]
best_model = trained_models[best_model_name]

print("\nBest ML Model:", best_model_name)

joblib.dump(
    best_model,
    ROOT / "coal_production_model.pkl"
)

with open(
    ROOT / "coal_production_model_pickle.pkl",
    "wb"
) as file:
    pickle.dump(best_model, file)

# ================================================================
# 10. PREPARE DATA FOR NEURAL NETWORK
# ================================================================

nn_preprocessor = make_preprocessor()

X_train_nn = nn_preprocessor.fit_transform(X_train)
X_test_nn = nn_preprocessor.transform(X_test)

target_scaler = StandardScaler()

y_train_nn = target_scaler.fit_transform(
    y_train.to_numpy().reshape(-1, 1)
).ravel()

joblib.dump(
    nn_preprocessor,
    ROOT / "nn_preprocessor.pkl"
)

joblib.dump(
    target_scaler,
    ROOT / "target_scaler.pkl"
)

# ================================================================
# 11. BUILD ARTIFICIAL NEURAL NETWORK
# ================================================================

ann_model = Sequential([
    Input(shape=(X_train_nn.shape[1],)),

    Dense(64, activation="relu"),
    Dropout(0.10),

    Dense(32, activation="relu"),

    Dense(16, activation="relu"),

    Dense(1, activation="linear")
])

ann_model.compile(
    optimizer=Adam(learning_rate=0.001),
    loss="mse",
    metrics=["mae"]
)

early_stopping = EarlyStopping(
    monitor="val_loss",
    patience=20,
    restore_best_weights=True
)

history = ann_model.fit(
    X_train_nn,
    y_train_nn,
    validation_split=0.20,
    epochs=200,
    batch_size=16,
    callbacks=[early_stopping],
    verbose=0
)

# ================================================================
# 12. SAVE NEURAL NETWORK AS H5
# ================================================================

ann_model.save(
    str(ROOT / "coal_production_model.h5")
)

# Save native Keras format as well
ann_model.save(
    str(ROOT / "coal_production_model.keras")
)

print("\nNeural Network Models Saved.")

# ================================================================
# 13. EVALUATE NEURAL NETWORK
# ================================================================

ann_scaled_predictions = ann_model.predict(
    X_test_nn,
    verbose=0
)

ann_predictions = target_scaler.inverse_transform(
    ann_scaled_predictions
).ravel()

ann_metrics = calculate_metrics(
    y_test,
    ann_predictions
)

evaluation_results.append({
    "Model": "ANN",
    **ann_metrics
})

print("\nANN Performance:")
print(ann_metrics)

# ================================================================
# 14. SAVE MODEL COMPARISON
# ================================================================

comparison_df = pd.DataFrame(
    evaluation_results
).sort_values("RMSE")

comparison_df.to_csv(
    ROOT / "model_comparison.csv",
    index=False
)

print("\nModel Comparison:")
print(comparison_df)

# ================================================================
# 15. MODEL COMPARISON GRAPH
# ================================================================

plt.figure(figsize=(11, 6))

sns.barplot(
    data=comparison_df,
    x="Model",
    y="RMSE",
    color="cornflowerblue"
)

plt.title("Model Comparison Using RMSE")
plt.ylabel("RMSE (Lower Is Better)")
plt.xticks(rotation=30)
plt.tight_layout()

plt.savefig(
    ROOT / "comparison_graph.png",
    dpi=300
)

plt.show()

# ================================================================
# 16. NEURAL NETWORK TRAINING GRAPH
# ================================================================

plt.figure(figsize=(10, 6))

plt.plot(
    history.history["loss"],
    label="Training Loss"
)

plt.plot(
    history.history["val_loss"],
    label="Validation Loss"
)

plt.title("Neural Network Training Performance")
plt.xlabel("Epoch")
plt.ylabel("Scaled Mean Squared Error")
plt.legend()
plt.tight_layout()

plt.savefig(
    ROOT / "ann_training_graph.png",
    dpi=300
)

plt.show()

# ================================================================
# 17. GENERATE PREDICTIONS
# ================================================================

best_predictions = best_model.predict(X_test)

test_results = data.loc[X_test.index, [
    "Mine",
    "Company",
    "Coal_Production"
]].copy()

test_results["Predicted_Production"] = best_predictions

test_results["Absolute_Error"] = np.abs(
    test_results["Coal_Production"] -
    test_results["Predicted_Production"]
)

test_results["ANN_Predicted_Production"] = ann_predictions

test_results.to_csv(
    ROOT / "prediction.csv",
    index=False
)

test_results.to_csv(
    ROOT / "result.csv",
    index=False
)

# ================================================================
# 18. ACTUAL VS PREDICTED GRAPH
# ================================================================

plt.figure(figsize=(9, 7))

plt.scatter(
    y_test,
    best_predictions,
    alpha=0.7
)

lower = min(
    float(y_test.min()),
    float(np.min(best_predictions))
)

upper = max(
    float(y_test.max()),
    float(np.max(best_predictions))
)

plt.plot(
    [lower, upper],
    [lower, upper],
    "r--",
    label="Perfect Prediction"
)

plt.xlabel("Actual Production")
plt.ylabel("Predicted Production")
plt.title("Actual vs Predicted Coal Production")
plt.legend()
plt.tight_layout()

plt.savefig(
    ROOT / "prediction_graph.png",
    dpi=300
)

plt.show()

# ================================================================
# 19. PRODUCTION CLASSIFICATION
# ================================================================

# Descriptive categories based on dataset quantiles.
# These are not independently validated operational ratings.

data["Production_Category"] = pd.qcut(
    data["Coal_Production"].rank(method="first"),
    q=3,
    labels=["Low", "Medium", "High"]
)

data.to_csv(
    ROOT / "production_classification.csv",
    index=False
)

print("\nProduction Classification:")
print(data["Production_Category"].value_counts())

# ================================================================
# 20. CONFIGURATION YAML
# ================================================================

config = {
    "project_name": (
        "AI-Based Coal Mine Production Analysis, "
        "Mining Company Performance Assessment, "
        "and Production Optimization System"
    ),
    "dataset": CSV_PATH.name,
    "random_seed": SEED,
    "target_column": "Coal_Production",
    "feature_columns": ["Company"],
    "test_size": 0.20,
    "best_ml_model": best_model_name,
    "model_selection_metric": "RMSE",
    "neural_network": {
        "hidden_layers": [64, 32, 16],
        "activation": "relu",
        "optimizer": "adam",
        "loss": "mse",
        "max_epochs": 200,
        "batch_size": 16
    },
    "notes": (
        "Cross-sectional data supports descriptive analysis "
        "and limited predictive benchmarking, not validated "
        "future production forecasting."
    )
}

with open(
    ROOT / "config.yaml",
    "w",
    encoding="utf-8"
) as file:
    yaml.safe_dump(
        config,
        file,
        sort_keys=False,
        allow_unicode=True
    )

# ================================================================
# 21. MODEL METADATA JSON
# ================================================================

metadata = {
    "project": config["project_name"],
    "dataset": CSV_PATH.name,
    "total_records": int(len(data)),
    "number_of_companies": int(
        data["Company"].nunique()
    ),
    "number_of_mines": int(
        data["Mine"].nunique()
    ),
    "features": ["Company"],
    "target": "Coal_Production",
    "best_ml_model": best_model_name,
    "model_metrics": evaluation_results,
    "files": {
        "h5": "coal_production_model.h5",
        "keras": "coal_production_model.keras",
        "pkl": "coal_production_model.pkl",
        "yaml": "config.yaml",
        "json": "model_metadata.json"
    }
}

with open(
    ROOT / "model_metadata.json",
    "w",
    encoding="utf-8"
) as file:
    json.dump(
        metadata,
        file,
        indent=4,
        ensure_ascii=False
    )

# ================================================================
# 22. VERIFY SAVED FILES
# ================================================================

required_files = [
    "coal_production_model.h5",
    "coal_production_model.pkl",
    "config.yaml",
    "model_metadata.json"
]

print("\n" + "=" * 65)
print("PROJECT EXECUTION COMPLETED")
print("=" * 65)

for filename in required_files:
    path = ROOT / filename

    print(
        filename,
        "CREATED" if path.exists() else "MISSING"
    )

print("\nAll Generated Files:")

for path in sorted(ROOT.iterdir()):
    if path.is_file():
        print(path.name)

print("\nOutput Directory:")
print(ROOT)

print("\nBest ML Model:", best_model_name)
print("\nFinal Model Comparison:")
print(comparison_df.to_string(index=False))
```
