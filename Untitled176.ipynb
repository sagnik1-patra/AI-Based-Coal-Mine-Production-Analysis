```python
import os
import json
import yaml
import joblib
import warnings
from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import (
    OneHotEncoder,
    StandardScaler
)
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer

from sklearn.dummy import DummyRegressor
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import (
    RandomForestRegressor,
    GradientBoostingRegressor,
    ExtraTreesRegressor
)
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import (
    Input,
    Dense,
    Dropout
)
from tensorflow.keras.callbacks import EarlyStopping

warnings.filterwarnings("ignore", category=FutureWarning)

# ============================================================
# 1. PATH CONFIGURATION
# ============================================================

ROOT = Path(
    r"C:\Users\sagni\Downloads\AI-Based Coal Mine Production Analysis, Mining Company Performance Assessment, and Production Optimization System"
)

CSV_PATH = ROOT / "rs_session-241_au482_1.1.csv"

ROOT.mkdir(parents=True, exist_ok=True)

if not CSV_PATH.exists():
    raise FileNotFoundError(
        f"Dataset not found: {CSV_PATH}"
    )

SEED = 42
np.random.seed(SEED)
tf.random.set_seed(SEED)

sns.set_theme(style="whitegrid")

print("=" * 65)
print("COAL MINE PRODUCTION ANALYSIS SYSTEM")
print("=" * 65)

print("Dataset:", CSV_PATH)
print("Output Folder:", ROOT)

# ============================================================
# 2. LOAD DATASET
# ============================================================

try:
    df = pd.read_csv(
        CSV_PATH,
        encoding="utf-8-sig"
    )
except UnicodeDecodeError:
    df = pd.read_csv(
        CSV_PATH,
        encoding="latin1"
    )

df.columns = (
    df.columns.astype(str)
    .str.strip()
    .str.replace(r"\s+", " ", regex=True)
)

df = df.dropna(how="all").copy()

print("\nDataset Shape:", df.shape)
print("\nColumns:")
print(df.columns.tolist())

print("\nFirst Five Records:")
print(df.head())

# ============================================================
# 3. AUTOMATIC COLUMN DETECTION
# ============================================================

def find_column(columns, keywords):

    for keyword in keywords:
        for col in columns:
            if keyword.lower() in col.lower():
                return col

    return None


mine_col = find_column(
    df.columns,
    [
        "Name of the Mine",
        "Mine Name",
        "Mine"
    ]
)

company_col = find_column(
    df.columns,
    [
        "Type of WCL",
        "Company",
        "Coalfield"
    ]
)

target_col = find_column(
    df.columns,
    [
        "Quantity for Coal production",
        "Coal Production",
        "Quantity"
    ]
)

if not all([mine_col, company_col, target_col]):
    raise ValueError(
        "Required columns could not be identified.\n"
        f"Available columns: {df.columns.tolist()}"
    )

data = df[
    [mine_col, company_col, target_col]
].copy()

data.columns = [
    "Mine",
    "Company",
    "Coal_Production"
]

# ============================================================
# 4. DATA CLEANING
# ============================================================

for col in ["Mine", "Company"]:

    data[col] = (
        data[col]
        .astype("string")
        .str.strip()
        .replace("", pd.NA)
        .fillna("Unknown")
        .astype(str)
    )

data["Coal_Production"] = pd.to_numeric(
    data["Coal_Production"]
    .astype(str)
    .str.replace(",", "", regex=False)
    .str.strip(),
    errors="coerce"
)

data = data.dropna(
    subset=["Coal_Production"]
)

data = data[
    data["Coal_Production"] >= 0
]

data = data.drop_duplicates().reset_index(
    drop=True
)

if len(data) < 15:
    raise ValueError(
        "Not enough valid records for model evaluation."
    )

data.to_csv(
    ROOT / "cleaned_dataset.csv",
    index=False
)

print("\nClean Dataset Shape:", data.shape)

# ============================================================
# 5. COMPANY PERFORMANCE ANALYSIS
# ============================================================

company_summary = (
    data.groupby("Company")["Coal_Production"]
    .agg(
        Mine_Count="count",
        Total_Production="sum",
        Average_Production="mean",
        Median_Production="median",
        Production_Std="std"
    )
    .reset_index()
    .sort_values(
        "Total_Production",
        ascending=False
    )
)

company_summary.to_csv(
    ROOT / "company_performance.csv",
    index=False
)

print("\nCompany Performance:")
print(company_summary)

# ============================================================
# 6. FEATURE PREPARATION
# ============================================================

# Company is used as the predictive feature.
# Mine names are identifiers, so they are retained for
# reporting instead of being used to memorize records.

X = data[["Company"]].copy()

y = data["Coal_Production"].astype(float)

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=SEED
)

def make_preprocessor():

    try:
        encoder = OneHotEncoder(
            handle_unknown="ignore",
            sparse_output=False
        )
    except TypeError:
        encoder = OneHotEncoder(
            handle_unknown="ignore",
            sparse=False
        )

    categorical_pipeline = Pipeline([
        (
            "imputer",
            SimpleImputer(
                strategy="most_frequent"
            )
        ),
        (
            "encoder",
            encoder
        )
    ])

    return ColumnTransformer([
        (
            "categorical",
            categorical_pipeline,
            ["Company"]
        )
    ])

# ============================================================
# 7. MACHINE LEARNING MODELS
# ============================================================

models = {
    "Baseline": DummyRegressor(
        strategy="mean"
    ),

    "Decision Tree": DecisionTreeRegressor(
        max_depth=5,
        min_samples_leaf=3,
        random_state=SEED
    ),

    "Random Forest": RandomForestRegressor(
        n_estimators=200,
        max_depth=8,
        min_samples_leaf=2,
        random_state=SEED,
        n_jobs=-1
    ),

    "Gradient Boosting": GradientBoostingRegressor(
        n_estimators=150,
        learning_rate=0.05,
        max_depth=3,
        random_state=SEED
    ),

    "Extra Trees": ExtraTreesRegressor(
        n_estimators=200,
        max_depth=8,
        min_samples_leaf=2,
        random_state=SEED,
        n_jobs=-1
    )
}

trained_models = {}
results = []
predictions_by_model = {}

def evaluate_model(actual, predicted):

    return {
        "MAE": float(
            mean_absolute_error(actual, predicted)
        ),
        "MSE": float(
            mean_squared_error(actual, predicted)
        ),
        "RMSE": float(
            np.sqrt(
                mean_squared_error(actual, predicted)
            )
        ),
        "R2": float(
            r2_score(actual, predicted)
        )
    }

for name, estimator in models.items():

    pipeline = Pipeline([
        (
            "preprocessor",
            make_preprocessor()
        ),
        (
            "model",
            estimator
        )
    ])

    pipeline.fit(X_train, y_train)

    pred = pipeline.predict(X_test)

    metrics = evaluate_model(
        y_test,
        pred
    )

    results.append({
        "Model": name,
        **metrics
    })

    trained_models[name] = pipeline
    predictions_by_model[name] = pred

    print(f"\n{name}")
    print("MAE:", round(metrics["MAE"], 4))
    print("RMSE:", round(metrics["RMSE"], 4))
    print("R2:", round(metrics["R2"], 4))

# ============================================================
# 8. ARTIFICIAL NEURAL NETWORK
# ============================================================

nn_preprocessor = make_preprocessor()

X_train_nn = nn_preprocessor.fit_transform(
    X_train
)

X_test_nn = nn_preprocessor.transform(
    X_test
)

target_scaler = StandardScaler()

y_train_scaled = target_scaler.fit_transform(
    y_train.to_numpy().reshape(-1, 1)
).ravel()

ann = Sequential([
    Input(shape=(X_train_nn.shape[1],)),

    Dense(64, activation="relu"),
    Dropout(0.10),

    Dense(32, activation="relu"),

    Dense(16, activation="relu"),

    Dense(1, activation="linear")
])

ann.compile(
    optimizer="adam",
    loss="mse",
    metrics=["mae"]
)

early_stop = EarlyStopping(
    monitor="val_loss",
    patience=20,
    restore_best_weights=True
)

history = ann.fit(
    X_train_nn,
    y_train_scaled,
    validation_split=0.20,
    epochs=200,
    batch_size=16,
    callbacks=[early_stop],
    verbose=0
)

ann_pred_scaled = ann.predict(
    X_test_nn,
    verbose=0
)

ann_pred = target_scaler.inverse_transform(
    ann_pred_scaled
).ravel()

ann_metrics = evaluate_model(
    y_test,
    ann_pred
)

results.append({
    "Model": "ANN",
    **ann_metrics
})

predictions_by_model["ANN"] = ann_pred

ann.save(
    str(ROOT / "coal_production_model.h5")
)

ann.save(
    str(ROOT / "coal_production_model.keras")
)

joblib.dump(
    nn_preprocessor,
    ROOT / "nn_preprocessor.pkl"
)

joblib.dump(
    target_scaler,
    ROOT / "target_scaler.pkl"
)

# ============================================================
# 9. MODEL COMPARISON DATA
# ============================================================

comparison_df = pd.DataFrame(
    results
).sort_values(
    "RMSE",
    ascending=True
).reset_index(drop=True)

comparison_df.to_csv(
    ROOT / "model_comparison.csv",
    index=False
)

print("\nModel Comparison:")
print(comparison_df)

# Select best traditional ML model.
# The ANN is stored separately in H5 format.

ml_comparison = comparison_df[
    comparison_df["Model"] != "ANN"
]

best_model_name = ml_comparison.iloc[0]["Model"]

best_model = trained_models[
    best_model_name
]

joblib.dump(
    best_model,
    ROOT / "coal_production_model.pkl"
)

print("\nBest Traditional ML Model:", best_model_name)

# ============================================================
# 10. ACCURACY GRAPH
# ============================================================

# Regression uses R2, not classification accuracy.
# R2 can be negative and is not a percentage.

plt.figure(figsize=(12, 6))

ax = sns.barplot(
    data=comparison_df,
    x="Model",
    y="R2",
    color="royalblue"
)

plt.axhline(
    y=0,
    color="black",
    linestyle="--",
    linewidth=1
)

plt.title(
    "Regression Model Performance (R2 Score)",
    fontsize=15
)

plt.xlabel("Machine Learning Models")
plt.ylabel("R2 Score (Higher Is Better)")

plt.xticks(rotation=30, ha="right")

for container in ax.containers:
    ax.bar_label(
        container,
        fmt="%.3f",
        padding=3
    )

plt.tight_layout()

plt.savefig(
    ROOT / "accuracy_graph.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

print("Saved: accuracy_graph.png")

# ============================================================
# 11. CORRELATION HEATMAP
# ============================================================

# Numeric production data is combined with company-level
# descriptive statistics for an interpretable heatmap.
# This is descriptive, not evidence of causal relationships.

heatmap_data = data.merge(
    company_summary[
        [
            "Company",
            "Mine_Count",
            "Total_Production",
            "Average_Production",
            "Median_Production",
            "Production_Std"
        ]
    ],
    on="Company",
    how="left"
)

numeric_data = heatmap_data.select_dtypes(
    include=[np.number]
)

correlation_matrix = numeric_data.corr()

plt.figure(figsize=(11, 8))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    linewidths=0.5,
    vmin=-1,
    vmax=1
)

plt.title(
    "Coal Production and Company Statistics Correlation",
    fontsize=14
)

plt.tight_layout()

plt.savefig(
    ROOT / "heatmap.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

print("Saved: heatmap.png")

# ============================================================
# 12. MODEL COMPARISON GRAPH
# ============================================================

# Compare models using MAE and RMSE.
# Lower values indicate better regression performance.

comparison_plot = comparison_df.melt(
    id_vars="Model",
    value_vars=["MAE", "RMSE"],
    var_name="Metric",
    value_name="Value"
)

plt.figure(figsize=(13, 7))

sns.barplot(
    data=comparison_plot,
    x="Model",
    y="Value",
    hue="Metric"
)

plt.title(
    "Machine Learning Model Comparison",
    fontsize=15
)

plt.xlabel("Models")
plt.ylabel("Error Value (Lower Is Better)")
plt.xticks(rotation=30, ha="right")

plt.legend(title="Evaluation Metric")

plt.tight_layout()

plt.savefig(
    ROOT / "comparison_graph.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

print("Saved: comparison_graph.png")

# ============================================================
# 13. RESULT CSV
# ============================================================

best_pred = best_model.predict(
    X_test
)

result_df = data.loc[
    X_test.index,
    [
        "Mine",
        "Company",
        "Coal_Production"
    ]
].copy()

result_df["Predicted_Production"] = best_pred

result_df["ANN_Predicted_Production"] = ann_pred

result_df["Absolute_Error"] = np.abs(
    result_df["Coal_Production"] -
    result_df["Predicted_Production"]
)

result_df["Squared_Error"] = (
    result_df["Coal_Production"] -
    result_df["Predicted_Production"]
) ** 2

result_df["Percentage_Error"] = np.where(
    result_df["Coal_Production"] != 0,
    (
        result_df["Absolute_Error"] /
        result_df["Coal_Production"].replace(0, np.nan)
    ) * 100,
    np.nan
)

result_df = result_df.reset_index(
    drop=True
)

result_df.to_csv(
    ROOT / "result.csv",
    index=False
)

print("\nResult CSV Created:")
print(result_df.head())

# ============================================================
# 14. RESULT GRAPH
# ============================================================

# Plot the first 30 test records for readability.

plot_result = result_df.head(30)

x_positions = np.arange(
    len(plot_result)
)

plt.figure(figsize=(15, 7))

plt.plot(
    x_positions,
    plot_result["Coal_Production"],
    marker="o",
    label="Actual Production",
    linewidth=2
)

plt.plot(
    x_positions,
    plot_result["Predicted_Production"],
    marker="s",
    label="Predicted Production",
    linewidth=2
)

plt.title(
    "Actual vs Predicted Coal Production",
    fontsize=15
)

plt.xlabel("Test Record")
plt.ylabel("Coal Production")

plt.xticks(
    x_positions,
    rotation=45
)

plt.legend()
plt.tight_layout()

plt.savefig(
    ROOT / "result_graph.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

print("Saved: result_graph.png")

# ============================================================
# 15. PREDICTION CSV
# ============================================================

# Predictions for all observed records.
# These are model estimates, not future forecasts.
# The test rows were not used for model fitting.

all_predictions = best_model.predict(
    X
)

all_nn_features = nn_preprocessor.transform(
    X
)

all_ann_scaled = ann.predict(
    all_nn_features,
    verbose=0
)

all_ann_predictions = target_scaler.inverse_transform(
    all_ann_scaled
).ravel()

prediction_df = data[
    [
        "Mine",
        "Company",
        "Coal_Production"
    ]
].copy()

prediction_df["ML_Predicted_Production"] = (
    all_predictions
)

prediction_df["ANN_Predicted_Production"] = (
    all_ann_predictions
)

prediction_df["Data_Split"] = np.where(
    prediction_df.index.isin(X_test.index),
    "Test",
    "Train"
)

prediction_df.to_csv(
    ROOT / "prediction.csv",
    index=False
)

print("\nPrediction CSV Created:")
print(prediction_df.head())

# ============================================================
# 16. PREDICTION GRAPH
# ============================================================

# Use held-out test records for the evaluation graph.

plt.figure(figsize=(10, 8))

plt.scatter(
    y_test,
    best_pred,
    alpha=0.7,
    s=65,
    label=best_model_name
)

plt.scatter(
    y_test,
    ann_pred,
    alpha=0.5,
    s=50,
    marker="x",
    label="ANN"
)

minimum = min(
    float(y_test.min()),
    float(np.min(best_pred)),
    float(np.min(ann_pred))
)

maximum = max(
    float(y_test.max()),
    float(np.max(best_pred)),
    float(np.max(ann_pred))
)

plt.plot(
    [minimum, maximum],
    [minimum, maximum],
    "r--",
    linewidth=2,
    label="Perfect Prediction"
)

plt.title(
    "Coal Production Prediction Analysis",
    fontsize=15
)

plt.xlabel("Actual Coal Production")
plt.ylabel("Predicted Coal Production")

plt.legend()
plt.tight_layout()

plt.savefig(
    ROOT / "prediction_graph.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

print("Saved: prediction_graph.png")

# ============================================================
# 17. ADDITIONAL COMPANY PRODUCTION GRAPH
# ============================================================

plt.figure(figsize=(13, 7))

sns.barplot(
    data=company_summary,
    x="Company",
    y="Total_Production",
    color="teal"
)

plt.title(
    "Company-Wise Total Coal Production",
    fontsize=15
)

plt.xlabel("Mining Company")
plt.ylabel("Total Coal Production")

plt.xticks(rotation=45, ha="right")
plt.tight_layout()

plt.savefig(
    ROOT / "company_production_graph.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ============================================================
# 18. NEURAL NETWORK TRAINING GRAPH
# ============================================================

plt.figure(figsize=(11, 6))

plt.plot(
    history.history["loss"],
    label="Training Loss"
)

plt.plot(
    history.history["val_loss"],
    label="Validation Loss"
)

plt.title(
    "ANN Training and Validation Loss"
)

plt.xlabel("Epoch")
plt.ylabel("Scaled Mean Squared Error")

plt.legend()
plt.tight_layout()

plt.savefig(
    ROOT / "ann_training_graph.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ============================================================
# 19. SAVE YAML CONFIGURATION
# ============================================================

config = {
    "project": (
        "AI-Based Coal Mine Production Analysis, "
        "Mining Company Performance Assessment, "
        "and Production Optimization System"
    ),
    "dataset": CSV_PATH.name,
    "target": "Coal_Production",
    "features": ["Company"],
    "test_size": 0.20,
    "random_state": SEED,
    "best_ml_model": best_model_name,
    "evaluation_metrics": [
        "MAE",
        "MSE",
        "RMSE",
        "R2"
    ],
    "output_files": [
        "accuracy_graph.png",
        "heatmap.png",
        "comparison_graph.png",
        "result.csv",
        "result_graph.png",
        "prediction.csv",
        "prediction_graph.png"
    ]
}

with open(
    ROOT / "config.yaml",
    "w",
    encoding="utf-8"
) as file:

    yaml.safe_dump(
        config,
        file,
        sort_keys=False
    )

# ============================================================
# 20. SAVE JSON METADATA
# ============================================================

metadata = {
    "project": config["project"],
    "dataset": CSV_PATH.name,
    "total_records": int(len(data)),
    "total_companies": int(
        data["Company"].nunique()
    ),
    "total_mines": int(
        data["Mine"].nunique()
    ),
    "best_ml_model": best_model_name,
    "model_results": comparison_df.to_dict(
        orient="records"
    ),
    "notes": (
        "Results are based on company-only features. "
        "The dataset does not contain a time dimension "
        "for future production forecasting."
    )
}

with open(
    ROOT / "model_metadata.json",
    "w",
    encoding="utf-8"
) as file:

    json.dump(
        metadata,
        file,
        indent=4
    )

# ============================================================
# 21. VERIFY ALL OUTPUTS
# ============================================================

required_outputs = [
    "accuracy_graph.png",
    "heatmap.png",
    "comparison_graph.png",
    "result.csv",
    "result_graph.png",
    "prediction.csv",
    "prediction_graph.png",
    "coal_production_model.pkl",
    "coal_production_model.h5",
    "config.yaml",
    "model_metadata.json"
]

print("\n" + "=" * 65)
print("FINAL OUTPUT VERIFICATION")
print("=" * 65)

for filename in required_outputs:

    file_path = ROOT / filename

    if file_path.exists():
        print("[SUCCESS]", filename)
    else:
        print("[MISSING]", filename)

print("\n" + "=" * 65)
print("MODEL PERFORMANCE SUMMARY")
print("=" * 65)

print(
    comparison_df.to_string(
        index=False
    )
)

print("\nBest Traditional ML Model:", best_model_name)

print("\nAll Results Saved At:")
print(ROOT)

print("\nPROJECT EXECUTION COMPLETED SUCCESSFULLY!")
```
