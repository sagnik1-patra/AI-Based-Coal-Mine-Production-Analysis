```python
import os
import json
import yaml
import joblib
import warnings
from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import (
    train_test_split,
    KFold,
    cross_val_score
)

from sklearn.preprocessing import (
    OneHotEncoder,
    StandardScaler
)

from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer

from sklearn.ensemble import RandomForestRegressor

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

import tensorflow as tf

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import (
    Input,
    Dense,
    Dropout
)
from tensorflow.keras.callbacks import EarlyStopping

warnings.filterwarnings(
    "ignore",
    category=FutureWarning
)

# ================================================================
# 1. PROJECT CONFIGURATION
# ================================================================

ROOT = Path(
    r"C:\Users\sagni\Downloads\AI-Based Coal Mine Production Analysis, Mining Company Performance Assessment, and Production Optimization System"
)

CSV_PATH = ROOT / "rs_session-241_au482_1.1.csv"

ROOT.mkdir(
    parents=True,
    exist_ok=True
)

if not CSV_PATH.exists():
    raise FileNotFoundError(
        f"Dataset not found: {CSV_PATH}"
    )

SEED = 42

np.random.seed(SEED)
tf.random.set_seed(SEED)

rng = np.random.default_rng(SEED)

sns.set_theme(
    style="whitegrid"
)

print("=" * 70)
print("PARTICLE SWARM OPTIMIZATION")
print("COAL MINE PRODUCTION ANALYSIS")
print("=" * 70)

print("\nDataset:", CSV_PATH)
print("Output Directory:", ROOT)

# ================================================================
# 2. LOAD DATASET
# ================================================================

try:
    df = pd.read_csv(
        CSV_PATH,
        encoding="utf-8-sig"
    )
except UnicodeDecodeError:
    df = pd.read_csv(
        CSV_PATH,
        encoding="latin1"
    )

df.columns = (
    df.columns.astype(str)
    .str.strip()
    .str.replace(
        r"\s+",
        " ",
        regex=True
    )
)

df = df.dropna(
    how="all"
).copy()

print("\nDataset Shape:", df.shape)
print("\nColumns:", df.columns.tolist())
print("\nFirst Five Records:")
print(df.head())

# ================================================================
# 3. AUTOMATIC COLUMN IDENTIFICATION
# ================================================================

def find_column(columns, keywords):

    for keyword in keywords:
        for col in columns:

            if keyword.lower() in col.lower():
                return col

    return None


mine_col = find_column(
    df.columns,
    [
        "Name of the Mine",
        "Mine Name",
        "Mine"
    ]
)

company_col = find_column(
    df.columns,
    [
        "Type of WCL",
        "Company",
        "Coalfield"
    ]
)

target_col = find_column(
    df.columns,
    [
        "Quantity for Coal production",
        "Coal Production",
        "Quantity"
    ]
)

if not all([
    mine_col,
    company_col,
    target_col
]):
    raise ValueError(
        "Required dataset columns were not found.\n"
        f"Available columns: {df.columns.tolist()}"
    )

data = df[
    [
        mine_col,
        company_col,
        target_col
    ]
].copy()

data.columns = [
    "Mine",
    "Company",
    "Coal_Production"
]

# ================================================================
# 4. DATA CLEANING
# ================================================================

for col in [
    "Mine",
    "Company"
]:

    data[col] = (
        data[col]
        .astype("string")
        .str.strip()
        .replace("", pd.NA)
        .fillna("Unknown")
        .astype(str)
    )

data["Coal_Production"] = pd.to_numeric(
    data["Coal_Production"]
    .astype(str)
    .str.replace(
        ",",
        "",
        regex=False
    )
    .str.strip(),
    errors="coerce"
)

data = data.dropna(
    subset=["Coal_Production"]
)

data = data[
    data["Coal_Production"] >= 0
]

data = data.drop_duplicates().reset_index(
    drop=True
)

if len(data) < 20:
    raise ValueError(
        "Insufficient valid records for PSO evaluation."
    )

data.to_csv(
    ROOT / "pso_cleaned_dataset.csv",
    index=False
)

print("\nCleaned Dataset Shape:", data.shape)

# ================================================================
# 5. COMPANY PERFORMANCE ANALYSIS
# ================================================================

company_summary = (
    data.groupby("Company")["Coal_Production"]
    .agg(
        Mine_Count="count",
        Total_Production="sum",
        Average_Production="mean",
        Median_Production="median",
        Production_Std="std"
    )
    .reset_index()
    .sort_values(
        "Total_Production",
        ascending=False
    )
)

company_summary.to_csv(
    ROOT / "pso_company_performance.csv",
    index=False
)

print("\nCompany Performance:")
print(company_summary)

# ================================================================
# 6. TRAIN TEST SPLIT
# ================================================================

# Company is used as the predictive feature.
# Mine names are identifiers and are retained for reporting.

X = data[["Company"]].copy()

y = data[
    "Coal_Production"
].astype(float)

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=SEED
)

def make_preprocessor():

    try:
        encoder = OneHotEncoder(
            handle_unknown="ignore",
            sparse_output=False
        )

    except TypeError:
        encoder = OneHotEncoder(
            handle_unknown="ignore",
            sparse=False
        )

    return ColumnTransformer([
        (
            "company",
            Pipeline([
                (
                    "imputer",
                    SimpleImputer(
                        strategy="most_frequent"
                    )
                ),
                (
                    "encoder",
                    encoder
                )
            ]),
            ["Company"]
        )
    ])


def make_rf_pipeline(params):

    return Pipeline([
        (
            "preprocessor",
            make_preprocessor()
        ),
        (
            "regressor",
            RandomForestRegressor(
                n_estimators=int(
                    params["n_estimators"]
                ),
                max_depth=int(
                    params["max_depth"]
                ),
                min_samples_split=int(
                    params["min_samples_split"]
                ),
                min_samples_leaf=int(
                    params["min_samples_leaf"]
                ),
                max_features=float(
                    params["max_features"]
                ),
                random_state=SEED,
                n_jobs=1
            )
        )
    ])

# ================================================================
# 7. PSO SEARCH SPACE
# ================================================================

PARAM_NAMES = [
    "n_estimators",
    "max_depth",
    "min_samples_split",
    "min_samples_leaf",
    "max_features"
]

LOWER_BOUNDS = np.array([
    50,
    2,
    2,
    1,
    0.3
], dtype=float)

UPPER_BOUNDS = np.array([
    250,
    15,
    15,
    10,
    1.0
], dtype=float)

INTEGER_PARAMS = {
    "n_estimators",
    "max_depth",
    "min_samples_split",
    "min_samples_leaf"
}

DIMENSIONS = len(PARAM_NAMES)

def decode_particle(position):

    clipped = np.clip(
        position,
        LOWER_BOUNDS,
        UPPER_BOUNDS
    )

    params = {}

    for i, name in enumerate(PARAM_NAMES):

        value = clipped[i]

        if name in INTEGER_PARAMS:
            params[name] = int(
                round(value)
            )
        else:
            params[name] = float(value)

    return params


def parameter_key(params):

    return tuple(
        round(
            float(params[name]),
            5
        )
        for name in PARAM_NAMES
    )

# ================================================================
# 8. PSO FITNESS FUNCTION
# ================================================================

cv = KFold(
    n_splits=3,
    shuffle=True,
    random_state=SEED
)

fitness_cache = {}

def fitness_function(position):

    params = decode_particle(
        position
    )

    key = parameter_key(
        params
    )

    if key in fitness_cache:
        return fitness_cache[key]

    model = make_rf_pipeline(
        params
    )

    scores = cross_val_score(
        model,
        X_train,
        y_train,
        cv=cv,
        scoring="neg_mean_squared_error",
        n_jobs=1
    )

    rmse = float(
        np.sqrt(
            -np.mean(scores)
        )
    )

    fitness_cache[key] = rmse

    return rmse

# ================================================================
# 9. PARTICLE SWARM OPTIMIZATION
# ================================================================

NUM_PARTICLES = 12
MAX_ITERATIONS = 12

INERTIA_WEIGHT = 0.7

COGNITIVE_COEFFICIENT = 1.5
SOCIAL_COEFFICIENT = 1.5

positions = rng.uniform(
    LOWER_BOUNDS,
    UPPER_BOUNDS,
    size=(
        NUM_PARTICLES,
        DIMENSIONS
    )
)

velocities = np.zeros_like(
    positions
)

personal_best_positions = positions.copy()

personal_best_scores = np.array([
    fitness_function(position)
    for position in positions
])

best_index = int(
    np.argmin(
        personal_best_scores
    )
)

global_best_position = (
    personal_best_positions[
        best_index
    ].copy()
)

global_best_score = float(
    personal_best_scores[
        best_index
    ]
)

optimization_history = []

print("\nStarting PSO Optimization...\n")

for iteration in range(MAX_ITERATIONS):

    for i in range(NUM_PARTICLES):

        r1 = rng.random(
            DIMENSIONS
        )

        r2 = rng.random(
            DIMENSIONS
        )

        cognitive_velocity = (
            COGNITIVE_COEFFICIENT
            * r1
            * (
                personal_best_positions[i]
                - positions[i]
            )
        )

        social_velocity = (
            SOCIAL_COEFFICIENT
            * r2
            * (
                global_best_position
                - positions[i]
            )
        )

        velocities[i] = (
            INERTIA_WEIGHT
            * velocities[i]
            + cognitive_velocity
            + social_velocity
        )

        positions[i] = (
            positions[i]
            + velocities[i]
        )

        positions[i] = np.clip(
            positions[i],
            LOWER_BOUNDS,
            UPPER_BOUNDS
        )

        score = fitness_function(
            positions[i]
        )

        if score < personal_best_scores[i]:

            personal_best_scores[i] = score

            personal_best_positions[i] = (
                positions[i].copy()
            )

            if score < global_best_score:

                global_best_score = float(score)

                global_best_position = (
                    positions[i].copy()
                )

    optimization_history.append({
        "Iteration": iteration + 1,
        "Best_CV_RMSE": global_best_score,
        "Evaluated_Configurations": len(
            fitness_cache
        )
    })

    print(
        f"Iteration {iteration + 1:02d}/{MAX_ITERATIONS} "
        f"| Best CV RMSE: {global_best_score:.5f}"
    )

best_params = decode_particle(
    global_best_position
)

print("\nPSO Optimization Completed!")

print("\nBest Hyperparameters:")
print(best_params)

print("\nBest Cross-Validation RMSE:")
print(global_best_score)

history_df = pd.DataFrame(
    optimization_history
)

history_df.to_csv(
    ROOT / "pso_optimization_history.csv",
    index=False
)

# ================================================================
# 10. TRAIN BASELINE AND PSO MODELS
# ================================================================

baseline_params = {
    "n_estimators": 100,
    "max_depth": 8,
    "min_samples_split": 2,
    "min_samples_leaf": 1,
    "max_features": 1.0
}

baseline_model = make_rf_pipeline(
    baseline_params
)

baseline_model.fit(
    X_train,
    y_train
)

pso_model = make_rf_pipeline(
    best_params
)

pso_model.fit(
    X_train,
    y_train
)

baseline_pred = baseline_model.predict(
    X_test
)

pso_pred = pso_model.predict(
    X_test
)

# ================================================================
# 11. EVALUATION METRICS
# ================================================================

def calculate_metrics(actual, predicted):

    return {
        "MAE": float(
            mean_absolute_error(
                actual,
                predicted
            )
        ),
        "MSE": float(
            mean_squared_error(
                actual,
                predicted
            )
        ),
        "RMSE": float(
            np.sqrt(
                mean_squared_error(
                    actual,
                    predicted
                )
            )
        ),
        "R2": float(
            r2_score(
                actual,
                predicted
            )
        )
    }

baseline_metrics = calculate_metrics(
    y_test,
    baseline_pred
)

pso_metrics = calculate_metrics(
    y_test,
    pso_pred
)

print("\nBaseline Metrics:")
print(baseline_metrics)

print("\nPSO Metrics:")
print(pso_metrics)

# ================================================================
# 12. TRAIN ARTIFICIAL NEURAL NETWORK
# ================================================================

# This is a separate ANN comparison model.
# PSO optimizes the Random Forest, not the ANN.

nn_preprocessor = make_preprocessor()

X_train_nn = nn_preprocessor.fit_transform(
    X_train
)

X_test_nn = nn_preprocessor.transform(
    X_test
)

target_scaler = StandardScaler()

y_train_scaled = target_scaler.fit_transform(
    y_train.to_numpy().reshape(-1, 1)
).ravel()

ann = Sequential([
    Input(
        shape=(
            X_train_nn.shape[1],
        )
    ),

    Dense(
        64,
        activation="relu"
    ),

    Dropout(0.10),

    Dense(
        32,
        activation="relu"
    ),

    Dense(
        16,
        activation="relu"
    ),

    Dense(
        1,
        activation="linear"
    )
])

ann.compile(
    optimizer="adam",
    loss="mse",
    metrics=["mae"]
)

early_stopping = EarlyStopping(
    monitor="val_loss",
    patience=20,
    restore_best_weights=True
)

ann_history = ann.fit(
    X_train_nn,
    y_train_scaled,
    validation_split=0.20,
    epochs=200,
    batch_size=16,
    callbacks=[early_stopping],
    verbose=0
)

ann_scaled_pred = ann.predict(
    X_test_nn,
    verbose=0
)

ann_pred = target_scaler.inverse_transform(
    ann_scaled_pred
).ravel()

ann_metrics = calculate_metrics(
    y_test,
    ann_pred
)

# ================================================================
# 13. SAVE MODELS
# ================================================================

joblib.dump(
    pso_model,
    ROOT / "pso_model.pkl"
)

joblib.dump(
    baseline_model,
    ROOT / "pso_baseline_model.pkl"
)

ann.save(
    str(ROOT / "pso_model.h5")
)

ann.save(
    str(ROOT / "pso_model.keras")
)

joblib.dump(
    nn_preprocessor,
    ROOT / "pso_nn_preprocessor.pkl"
)

joblib.dump(
    target_scaler,
    ROOT / "pso_target_scaler.pkl"
)

print("\nPSO Models Saved Successfully.")

# ================================================================
# 14. MODEL COMPARISON
# ================================================================

comparison_df = pd.DataFrame([
    {
        "Model": "Baseline RF",
        **baseline_metrics
    },
    {
        "Model": "PSO Optimized RF",
        **pso_metrics
    },
    {
        "Model": "ANN",
        **ann_metrics
    }
])

comparison_df.to_csv(
    ROOT / "pso_model_comparison.csv",
    index=False
)

print("\nModel Comparison:")
print(comparison_df)

# ================================================================
# 15. PSO ACCURACY GRAPH
# ================================================================

# R2 is used for regression performance.
# It is not classification accuracy.

plt.figure(figsize=(11, 6))

ax = sns.barplot(
    data=comparison_df,
    x="Model",
    y="R2",
    color="royalblue"
)

plt.axhline(
    0,
    color="black",
    linestyle="--"
)

plt.title(
    "PSO Model Performance - R2 Score",
    fontsize=15
)

plt.xlabel("Model")
plt.ylabel("R2 Score (Higher Is Better)")

for container in ax.containers:

    ax.bar_label(
        container,
        fmt="%.3f",
        padding=3
    )

plt.tight_layout()

plt.savefig(
    ROOT / "pso_accuracy_graph.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

print("Saved: pso_accuracy_graph.png")

# ================================================================
# 16. PSO CORRELATION HEATMAP
# ================================================================

heatmap_data = data.merge(
    company_summary[
        [
            "Company",
            "Mine_Count",
            "Total_Production",
            "Average_Production",
            "Median_Production",
            "Production_Std"
        ]
    ],
    on="Company",
    how="left"
)

numeric_data = heatmap_data.select_dtypes(
    include=[np.number]
)

correlation_matrix = numeric_data.corr()

plt.figure(figsize=(11, 8))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    linewidths=0.5,
    vmin=-1,
    vmax=1
)

plt.title(
    "PSO Coal Production Correlation Heatmap",
    fontsize=15
)

plt.tight_layout()

plt.savefig(
    ROOT / "pso_heatmap.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

print("Saved: pso_heatmap.png")

# ================================================================
# 17. PSO COMPARISON GRAPH
# ================================================================

plot_comparison = comparison_df.melt(
    id_vars="Model",
    value_vars=[
        "MAE",
        "RMSE"
    ],
    var_name="Metric",
    value_name="Error"
)

plt.figure(figsize=(12, 7))

sns.barplot(
    data=plot_comparison,
    x="Model",
    y="Error",
    hue="Metric"
)

plt.title(
    "PSO vs Baseline vs ANN",
    fontsize=15
)

plt.xlabel("Model")
plt.ylabel("Error (Lower Is Better)")

plt.tight_layout()

plt.savefig(
    ROOT / "pso_comparison_graph.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

print("Saved: pso_comparison_graph.png")

# ================================================================
# 18. PSO RESULT CSV
# ================================================================

result_df = data.loc[
    X_test.index,
    [
        "Mine",
        "Company",
        "Coal_Production"
    ]
].copy()

result_df["Baseline_Prediction"] = (
    baseline_pred
)

result_df["PSO_Prediction"] = (
    pso_pred
)

result_df["ANN_Prediction"] = (
    ann_pred
)

result_df["PSO_Absolute_Error"] = np.abs(
    result_df["Coal_Production"]
    - result_df["PSO_Prediction"]
)

result_df["PSO_Squared_Error"] = (
    result_df["Coal_Production"]
    - result_df["PSO_Prediction"]
) ** 2

result_df["PSO_Percentage_Error"] = np.where(
    result_df["Coal_Production"] != 0,
    (
        result_df["PSO_Absolute_Error"]
        / result_df["Coal_Production"].replace(
            0,
            np.nan
        )
    ) * 100,
    np.nan
)

result_df = result_df.reset_index(
    drop=True
)

result_df.to_csv(
    ROOT / "pso_result.csv",
    index=False
)

print("\nPSO Result CSV:")
print(result_df.head())

# ================================================================
# 19. PSO RESULT GRAPH
# ================================================================

plot_data = result_df.head(30)

positions_plot = np.arange(
    len(plot_data)
)

plt.figure(figsize=(15, 7))

plt.plot(
    positions_plot,
    plot_data["Coal_Production"],
    marker="o",
    linewidth=2,
    label="Actual Production"
)

plt.plot(
    positions_plot,
    plot_data["PSO_Prediction"],
    marker="s",
    linewidth=2,
    label="PSO Prediction"
)

plt.plot(
    positions_plot,
    plot_data["Baseline_Prediction"],
    linestyle="--",
    label="Baseline Prediction"
)

plt.title(
    "PSO Actual vs Predicted Coal Production",
    fontsize=15
)

plt.xlabel("Test Record")
plt.ylabel("Coal Production")

plt.legend()
plt.tight_layout()

plt.savefig(
    ROOT / "pso_result_graph.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

print("Saved: pso_result_graph.png")

# ================================================================
# 20. PSO PREDICTION CSV
# ================================================================

# Estimates for all observed records.
# These are not future production forecasts.

all_pso_predictions = pso_model.predict(
    X
)

all_baseline_predictions = baseline_model.predict(
    X
)

all_nn_features = nn_preprocessor.transform(
    X
)

all_ann_predictions = target_scaler.inverse_transform(
    ann.predict(
        all_nn_features,
        verbose=0
    )
).ravel()

prediction_df = data[
    [
        "Mine",
        "Company",
        "Coal_Production"
    ]
].copy()

prediction_df["PSO_Predicted_Production"] = (
    all_pso_predictions
)

prediction_df["Baseline_Predicted_Production"] = (
    all_baseline_predictions
)

prediction_df["ANN_Predicted_Production"] = (
    all_ann_predictions
)

prediction_df["Data_Split"] = np.where(
    prediction_df.index.isin(
        X_test.index
    ),
    "Test",
    "Train"
)

prediction_df.to_csv(
    ROOT / "pso_prediction.csv",
    index=False
)

print("\nPSO Prediction CSV:")
print(prediction_df.head())

# ================================================================
# 21. PSO PREDICTION GRAPH
# ================================================================

plt.figure(figsize=(10, 8))

plt.scatter(
    y_test,
    pso_pred,
    alpha=0.75,
    s=65,
    label="PSO Optimized RF"
)

plt.scatter(
    y_test,
    baseline_pred,
    alpha=0.45,
    s=45,
    marker="x",
    label="Baseline RF"
)

minimum = min(
    float(y_test.min()),
    float(np.min(pso_pred)),
    float(np.min(baseline_pred))
)

maximum = max(
    float(y_test.max()),
    float(np.max(pso_pred)),
    float(np.max(baseline_pred))
)

plt.plot(
    [
        minimum,
        maximum
    ],
    [
        minimum,
        maximum
    ],
    "r--",
    label="Perfect Prediction"
)

plt.title(
    "PSO Coal Production Prediction Graph",
    fontsize=15
)

plt.xlabel("Actual Coal Production")
plt.ylabel("Predicted Coal Production")

plt.legend()
plt.tight_layout()

plt.savefig(
    ROOT / "pso_prediction_graph.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

print("Saved: pso_prediction_graph.png")

# ================================================================
# 22. PSO CONVERGENCE GRAPH
# ================================================================

plt.figure(figsize=(11, 6))

plt.plot(
    history_df["Iteration"],
    history_df["Best_CV_RMSE"],
    marker="o",
    linewidth=2,
    label="Best CV RMSE"
)

plt.title(
    "PSO Optimization Convergence",
    fontsize=15
)

plt.xlabel("Iteration")
plt.ylabel("Best CV RMSE (Lower Is Better)")

plt.legend()
plt.tight_layout()

plt.savefig(
    ROOT / "pso_convergence_graph.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

print("Saved: pso_convergence_graph.png")

# ================================================================
# 23. ANN TRAINING GRAPH
# ================================================================

plt.figure(figsize=(11, 6))

plt.plot(
    ann_history.history["loss"],
    label="Training Loss"
)

plt.plot(
    ann_history.history["val_loss"],
    label="Validation Loss"
)

plt.title(
    "PSO Project - ANN Training History",
    fontsize=15
)

plt.xlabel("Epoch")
plt.ylabel("Scaled Mean Squared Error")

plt.legend()
plt.tight_layout()

plt.savefig(
    ROOT / "pso_ann_training_graph.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ================================================================
# 24. COMPANY PRODUCTION GRAPH
# ================================================================

plt.figure(figsize=(13, 7))

sns.barplot(
    data=company_summary,
    x="Company",
    y="Total_Production",
    color="teal"
)

plt.title(
    "PSO Company-Wise Coal Production Analysis",
    fontsize=15
)

plt.xlabel("Mining Company")
plt.ylabel("Total Coal Production")

plt.xticks(
    rotation=45,
    ha="right"
)

plt.tight_layout()

plt.savefig(
    ROOT / "pso_company_production_graph.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ================================================================
# 25. SAVE PSO CONFIGURATION
# ================================================================

config = {
    "project": (
        "AI-Based Coal Mine Production Analysis, "
        "Mining Company Performance Assessment, "
        "and Production Optimization System"
    ),
    "algorithm": "Particle Swarm Optimization",
    "dataset": CSV_PATH.name,
    "target": "Coal_Production",
    "features": ["Company"],
    "number_of_particles": NUM_PARTICLES,
    "maximum_iterations": MAX_ITERATIONS,
    "inertia_weight": INERTIA_WEIGHT,
    "cognitive_coefficient": COGNITIVE_COEFFICIENT,
    "social_coefficient": SOCIAL_COEFFICIENT,
    "fitness_metric": "Cross-validation RMSE",
    "best_hyperparameters": best_params,
    "best_cv_rmse": float(global_best_score),
    "test_size": 0.20,
    "random_seed": SEED
}

with open(
    ROOT / "pso_config.yaml",
    "w",
    encoding="utf-8"
) as file:

    yaml.safe_dump(
        config,
        file,
        sort_keys=False
    )

# ================================================================
# 26. SAVE PSO JSON METADATA
# ================================================================

metadata = {
    "project": config["project"],
    "algorithm": "Particle Swarm Optimization",
    "dataset": CSV_PATH.name,
    "records": int(len(data)),
    "companies": int(
        data["Company"].nunique()
    ),
    "mines": int(
        data["Mine"].nunique()
    ),
    "best_hyperparameters": best_params,
    "best_cross_validation_rmse": float(
        global_best_score
    ),
    "baseline_metrics": baseline_metrics,
    "pso_metrics": pso_metrics,
    "ann_metrics": ann_metrics,
    "evaluated_configurations": len(
        fitness_cache
    ),
    "notes": (
        "PSO optimizes Random Forest hyperparameters. "
        "The H5 model is a separate ANN comparison model. "
        "The dataset supports cross-sectional analysis, "
        "not validated future production forecasting."
    )
}

with open(
    ROOT / "pso_model_metadata.json",
    "w",
    encoding="utf-8"
) as file:

    json.dump(
        metadata,
        file,
        indent=4
    )

# ================================================================
# 27. VERIFY ALL GENERATED OUTPUTS
# ================================================================

required_files = [
    "pso_model.pkl",
    "pso_model.h5",
    "pso_config.yaml",
    "pso_model_metadata.json",
    "pso_accuracy_graph.png",
    "pso_heatmap.png",
    "pso_comparison_graph.png",
    "pso_result.csv",
    "pso_result_graph.png",
    "pso_prediction.csv",
    "pso_prediction_graph.png",
    "pso_convergence_graph.png",
    "pso_optimization_history.csv",
    "pso_company_performance.csv",
    "pso_company_production_graph.png",
    "pso_ann_training_graph.png"
]

print("\n" + "=" * 70)
print("PSO OUTPUT VERIFICATION")
print("=" * 70)

for filename in required_files:

    path = ROOT / filename

    if path.exists():
        print("[CREATED]", filename)
    else:
        print("[MISSING]", filename)

print("\n" + "=" * 70)
print("FINAL PSO PERFORMANCE")
print("=" * 70)

print(
    comparison_df.to_string(
        index=False
    )
)

print("\nBest PSO Parameters:")
print(best_params)

print("\nBest Cross-Validation RMSE:")
print(global_best_score)

print("\nAll PSO Files Saved In:")
print(ROOT)

print("\nPSO PROJECT EXECUTION COMPLETED!")
```
