```python
import os
import json
import yaml
import joblib
import warnings
from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import (
    train_test_split,
    KFold,
    cross_val_score
)
from sklearn.preprocessing import (
    OneHotEncoder,
    StandardScaler
)
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.ensemble import RandomForestRegressor
from sklearn.dummy import DummyRegressor
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Input, Dense, Dropout
from tensorflow.keras.callbacks import EarlyStopping

warnings.filterwarnings("ignore", category=FutureWarning)

# ================================================================
# 1. PROJECT PATH
# ================================================================

ROOT = Path(
    r"C:\Users\sagni\Downloads\AI-Based Coal Mine Production Analysis, Mining Company Performance Assessment, and Production Optimization System"
)

CSV_PATH = ROOT / "rs_session-241_au482_1.1.csv"

ROOT.mkdir(parents=True, exist_ok=True)

if not CSV_PATH.exists():
    raise FileNotFoundError(
        f"Dataset not found: {CSV_PATH}"
    )

SEED = 42

np.random.seed(SEED)
tf.random.set_seed(SEED)

sns.set_theme(style="whitegrid")

print("=" * 70)
print("ARTIFICIAL IMMUNE SYSTEM - COAL PRODUCTION ANALYSIS")
print("=" * 70)

# ================================================================
# 2. LOAD DATASET
# ================================================================

try:
    df = pd.read_csv(CSV_PATH, encoding="utf-8-sig")
except UnicodeDecodeError:
    df = pd.read_csv(CSV_PATH, encoding="latin1")

df.columns = (
    df.columns.astype(str)
    .str.strip()
    .str.replace(r"\s+", " ", regex=True)
)

df = df.dropna(how="all")

print("\nDataset Shape:", df.shape)
print("\nColumns:", df.columns.tolist())

# ================================================================
# 3. AUTOMATIC COLUMN IDENTIFICATION
# ================================================================

def find_column(columns, keywords):
    for keyword in keywords:
        for col in columns:
            if keyword.lower() in col.lower():
                return col
    return None

mine_col = find_column(
    df.columns,
    ["Name of the Mine", "Mine Name", "Mine"]
)

company_col = find_column(
    df.columns,
    ["Type of WCL", "Company", "Coalfield"]
)

target_col = find_column(
    df.columns,
    ["Quantity for Coal production", "Coal Production", "Quantity"]
)

if not all([mine_col, company_col, target_col]):
    raise ValueError(
        "Required columns not found. "
        f"Available columns: {df.columns.tolist()}"
    )

data = df[
    [mine_col, company_col, target_col]
].copy()

data.columns = [
    "Mine",
    "Company",
    "Coal_Production"
]

# ================================================================
# 4. CLEAN DATA
# ================================================================

for col in ["Mine", "Company"]:
    data[col] = (
        data[col]
        .astype("string")
        .str.strip()
        .replace("", pd.NA)
        .fillna("Unknown")
        .astype(str)
    )

data["Coal_Production"] = pd.to_numeric(
    data["Coal_Production"]
    .astype(str)
    .str.replace(",", "", regex=False)
    .str.strip(),
    errors="coerce"
)

data = data.dropna(subset=["Coal_Production"])
data = data[data["Coal_Production"] >= 0]
data = data.drop_duplicates().reset_index(drop=True)

if len(data) < 20:
    raise ValueError("Insufficient records for AIS evaluation.")

data.to_csv(
    ROOT / "ais_cleaned_dataset.csv",
    index=False
)

print("\nClean Dataset:", data.shape)

# ================================================================
# 5. COMPANY PERFORMANCE ANALYSIS
# ================================================================

company_summary = (
    data.groupby("Company")["Coal_Production"]
    .agg(
        Mine_Count="count",
        Total_Production="sum",
        Average_Production="mean",
        Median_Production="median",
        Production_Std="std"
    )
    .reset_index()
    .sort_values("Total_Production", ascending=False)
)

company_summary.to_csv(
    ROOT / "ais_company_performance.csv",
    index=False
)

print("\nCompany Performance:")
print(company_summary)

# ================================================================
# 6. TRAIN / TEST SPLIT
# ================================================================

# Company is the predictive feature.
# Mine names are retained for reporting, not model fitting.

X = data[["Company"]].copy()
y = data["Coal_Production"].astype(float)

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=SEED
)

def make_preprocessor():

    try:
        encoder = OneHotEncoder(
            handle_unknown="ignore",
            sparse_output=False
        )
    except TypeError:
        encoder = OneHotEncoder(
            handle_unknown="ignore",
            sparse=False
        )

    return ColumnTransformer([
        (
            "company",
            Pipeline([
                (
                    "imputer",
                    SimpleImputer(strategy="most_frequent")
                ),
                (
                    "encoder",
                    encoder
                )
            ]),
            ["Company"]
        )
    ])

def make_rf_pipeline(params):

    return Pipeline([
        (
            "preprocessor",
            make_preprocessor()
        ),
        (
            "regressor",
            RandomForestRegressor(
                n_estimators=int(params["n_estimators"]),
                max_depth=int(params["max_depth"]),
                min_samples_split=int(
                    params["min_samples_split"]
                ),
                min_samples_leaf=int(
                    params["min_samples_leaf"]
                ),
                max_features=float(
                    params["max_features"]
                ),
                random_state=SEED,
                n_jobs=1
            )
        )
    ])

# ================================================================
# 7. AIS SEARCH SPACE
# ================================================================

PARAM_BOUNDS = {
    "n_estimators": (50, 250),
    "max_depth": (2, 15),
    "min_samples_split": (2, 15),
    "min_samples_leaf": (1, 10),
    "max_features": (0.3, 1.0)
}

PARAM_NAMES = list(PARAM_BOUNDS.keys())

INTEGER_PARAMS = {
    "n_estimators",
    "max_depth",
    "min_samples_split",
    "min_samples_leaf"
}

rng = np.random.default_rng(SEED)

def random_antibody():

    antibody = {}

    for name, (low, high) in PARAM_BOUNDS.items():

        if name in INTEGER_PARAMS:
            antibody[name] = int(
                rng.integers(low, high + 1)
            )
        else:
            antibody[name] = float(
                rng.uniform(low, high)
            )

    return antibody

def clip_antibody(antibody):

    clipped = {}

    for name, value in antibody.items():

        low, high = PARAM_BOUNDS[name]
        value = np.clip(value, low, high)

        if name in INTEGER_PARAMS:
            value = int(round(value))
        else:
            value = float(value)

        clipped[name] = value

    return clipped

def antibody_key(antibody):

    return tuple(
        round(float(antibody[name]), 5)
        for name in PARAM_NAMES
    )

# ================================================================
# 8. FITNESS FUNCTION
# ================================================================

cv = KFold(
    n_splits=3,
    shuffle=True,
    random_state=SEED
)

fitness_cache = {}

def evaluate_antibody(antibody):

    antibody = clip_antibody(antibody)

    key = antibody_key(antibody)

    if key in fitness_cache:
        return fitness_cache[key]

    model = make_rf_pipeline(antibody)

    scores = cross_val_score(
        model,
        X_train,
        y_train,
        cv=cv,
        scoring="neg_mean_squared_error",
        n_jobs=1
    )

    # Root of mean cross-validation MSE.
    # Lower is better.
    rmse = float(
        np.sqrt(-np.mean(scores))
    )

    fitness_cache[key] = rmse

    return rmse

# ================================================================
# 9. CLONAL SELECTION AND HYPERMUTATION
# ================================================================

def mutate_antibody(antibody, mutation_strength):

    mutated = antibody.copy()

    for name in PARAM_NAMES:

        if rng.random() < 0.65:

            low, high = PARAM_BOUNDS[name]

            step = rng.normal(
                0,
                mutation_strength * (high - low)
            )

            mutated[name] = (
                mutated[name] + step
            )

    return clip_antibody(mutated)

def clone_and_mutate(antibody, rank, total_selected):

    # Better antibodies receive more clones.
    # Better antibodies also receive smaller mutations.

    clone_count = max(
        2,
        int(
            8 * (total_selected - rank) /
            total_selected
        )
    )

    mutation_strength = (
        0.08 + 0.30 * rank / total_selected
    )

    clones = []

    for _ in range(clone_count):

        clone = mutate_antibody(
            antibody,
            mutation_strength
        )

        clones.append(clone)

    return clones

# ================================================================
# 10. AIS OPTIMIZATION
# ================================================================

POPULATION_SIZE = 12
GENERATIONS = 12
SELECTED_COUNT = 5
RANDOM_REPLACEMENTS = 3

population = [
    random_antibody()
    for _ in range(POPULATION_SIZE)
]

best_antibody = None
best_fitness = float("inf")

optimization_history = []

print("\nStarting AIS Optimization...\n")

for generation in range(GENERATIONS):

    scored_population = [
        (
            evaluate_antibody(antibody),
            antibody
        )
        for antibody in population
    ]

    scored_population.sort(
        key=lambda item: item[0]
    )

    selected = scored_population[
        :SELECTED_COUNT
    ]

    clone_population = []

    for rank, (_, antibody) in enumerate(selected):

        clone_population.extend(
            clone_and_mutate(
                antibody,
                rank,
                SELECTED_COUNT
            )
        )

    clone_scores = [
        (
            evaluate_antibody(clone),
            clone
        )
        for clone in clone_population
    ]

    combined = scored_population + clone_scores

    combined.sort(
        key=lambda item: item[0]
    )

    if combined[0][0] < best_fitness:

        best_fitness = combined[0][0]

        best_antibody = combined[0][1].copy()

    # Keep high-affinity antibodies and inject diversity.
    survivors = [
        antibody.copy()
        for _, antibody in combined[
            :POPULATION_SIZE - RANDOM_REPLACEMENTS
        ]
    ]

    population = survivors + [
        random_antibody()
        for _ in range(RANDOM_REPLACEMENTS)
    ]

    optimization_history.append({
        "Generation": generation + 1,
        "Best_CV_RMSE": best_fitness,
        "Generation_Best_RMSE": combined[0][0],
        "Evaluated_Configurations": len(fitness_cache)
    })

    print(
        f"Generation {generation + 1:02d}/{GENERATIONS} "
        f"| Best CV RMSE: {best_fitness:.5f}"
    )

print("\nAIS Optimization Complete!")
print("\nBest Hyperparameters:")
print(best_antibody)

print("\nBest Cross-Validation RMSE:")
print(best_fitness)

history_df = pd.DataFrame(
    optimization_history
)

history_df.to_csv(
    ROOT / "ais_optimization_history.csv",
    index=False
)

# ================================================================
# 11. TRAIN BASELINE AND AIS-OPTIMIZED MODELS
# ================================================================

baseline_params = {
    "n_estimators": 100,
    "max_depth": 8,
    "min_samples_split": 2,
    "min_samples_leaf": 1,
    "max_features": 1.0
}

baseline_model = make_rf_pipeline(
    baseline_params
)

baseline_model.fit(
    X_train,
    y_train
)

ais_model = make_rf_pipeline(
    best_antibody
)

ais_model.fit(
    X_train,
    y_train
)

baseline_pred = baseline_model.predict(
    X_test
)

ais_pred = ais_model.predict(
    X_test
)

# ================================================================
# 12. EVALUATION METRICS
# ================================================================

def calculate_metrics(actual, predicted):

    return {
        "MAE": float(
            mean_absolute_error(actual, predicted)
        ),
        "MSE": float(
            mean_squared_error(actual, predicted)
        ),
        "RMSE": float(
            np.sqrt(
                mean_squared_error(actual, predicted)
            )
        ),
        "R2": float(
            r2_score(actual, predicted)
        )
    }

baseline_metrics = calculate_metrics(
    y_test,
    baseline_pred
)

ais_metrics = calculate_metrics(
    y_test,
    ais_pred
)

print("\nBaseline Metrics:")
print(baseline_metrics)

print("\nAIS Metrics:")
print(ais_metrics)

# ================================================================
# 13. TRAIN A SEPARATE NEURAL NETWORK
# ================================================================

# The ANN provides a comparison model and an H5 artifact.
# It is NOT optimized by the AIS search above.

nn_preprocessor = make_preprocessor()

X_train_nn = nn_preprocessor.fit_transform(
    X_train
)

X_test_nn = nn_preprocessor.transform(
    X_test
)

target_scaler = StandardScaler()

y_train_scaled = target_scaler.fit_transform(
    y_train.to_numpy().reshape(-1, 1)
).ravel()

ann = Sequential([
    Input(shape=(X_train_nn.shape[1],)),
    Dense(64, activation="relu"),
    Dropout(0.10),
    Dense(32, activation="relu"),
    Dense(16, activation="relu"),
    Dense(1, activation="linear")
])

ann.compile(
    optimizer="adam",
    loss="mse",
    metrics=["mae"]
)

early_stopping = EarlyStopping(
    monitor="val_loss",
    patience=20,
    restore_best_weights=True
)

ann_history = ann.fit(
    X_train_nn,
    y_train_scaled,
    validation_split=0.20,
    epochs=200,
    batch_size=16,
    callbacks=[early_stopping],
    verbose=0
)

ann_pred_scaled = ann.predict(
    X_test_nn,
    verbose=0
)

ann_pred = target_scaler.inverse_transform(
    ann_pred_scaled
).ravel()

ann_metrics = calculate_metrics(
    y_test,
    ann_pred
)

# ================================================================
# 14. SAVE MODELS
# ================================================================

joblib.dump(
    ais_model,
    ROOT / "ais_model.pkl"
)

joblib.dump(
    baseline_model,
    ROOT / "ais_baseline_model.pkl"
)

ann.save(
    str(ROOT / "ais_model.h5")
)

ann.save(
    str(ROOT / "ais_model.keras")
)

joblib.dump(
    nn_preprocessor,
    ROOT / "ais_nn_preprocessor.pkl"
)

joblib.dump(
    target_scaler,
    ROOT / "ais_target_scaler.pkl"
)

print("\nAIS Models Saved.")

# ================================================================
# 15. MODEL COMPARISON
# ================================================================

comparison_df = pd.DataFrame([
    {
        "Model": "Baseline RF",
        **baseline_metrics
    },
    {
        "Model": "AIS Optimized RF",
        **ais_metrics
    },
    {
        "Model": "ANN",
        **ann_metrics
    }
])

comparison_df.to_csv(
    ROOT / "ais_model_comparison.csv",
    index=False
)

print("\nModel Comparison:")
print(comparison_df)

# ================================================================
# 16. AIS ACCURACY GRAPH (R2)
# ================================================================

plt.figure(figsize=(11, 6))

ax = sns.barplot(
    data=comparison_df,
    x="Model",
    y="R2",
    color="royalblue"
)

plt.axhline(
    0,
    color="black",
    linestyle="--"
)

plt.title(
    "AIS Model Performance - R2 Score",
    fontsize=15
)

plt.ylabel("R2 Score (Higher Is Better)")
plt.xlabel("Model")

for container in ax.containers:
    ax.bar_label(
        container,
        fmt="%.3f",
        padding=3
    )

plt.tight_layout()

plt.savefig(
    ROOT / "ais_accuracy_graph.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ================================================================
# 17. AIS CORRELATION HEATMAP
# ================================================================

heatmap_data = data.merge(
    company_summary[
        [
            "Company",
            "Mine_Count",
            "Total_Production",
            "Average_Production",
            "Median_Production",
            "Production_Std"
        ]
    ],
    on="Company",
    how="left"
)

numeric_data = heatmap_data.select_dtypes(
    include=[np.number]
)

correlation_matrix = numeric_data.corr()

plt.figure(figsize=(11, 8))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    linewidths=0.5,
    vmin=-1,
    vmax=1
)

plt.title(
    "AIS Coal Production Correlation Heatmap",
    fontsize=15
)

plt.tight_layout()

plt.savefig(
    ROOT / "ais_heatmap.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ================================================================
# 18. AIS COMPARISON GRAPH
# ================================================================

plot_comparison = comparison_df.melt(
    id_vars="Model",
    value_vars=["MAE", "RMSE"],
    var_name="Metric",
    value_name="Error"
)

plt.figure(figsize=(12, 7))

sns.barplot(
    data=plot_comparison,
    x="Model",
    y="Error",
    hue="Metric"
)

plt.title(
    "AIS vs Baseline vs ANN",
    fontsize=15
)

plt.ylabel("Error (Lower Is Better)")
plt.xlabel("Model")

plt.tight_layout()

plt.savefig(
    ROOT / "ais_comparison_graph.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ================================================================
# 19. AIS RESULT CSV
# ================================================================

result_df = data.loc[
    X_test.index,
    [
        "Mine",
        "Company",
        "Coal_Production"
    ]
].copy()

result_df["Baseline_Prediction"] = baseline_pred
result_df["AIS_Prediction"] = ais_pred
result_df["ANN_Prediction"] = ann_pred

result_df["AIS_Absolute_Error"] = np.abs(
    result_df["Coal_Production"] -
    result_df["AIS_Prediction"]
)

result_df["AIS_Squared_Error"] = (
    result_df["Coal_Production"] -
    result_df["AIS_Prediction"]
) ** 2

result_df["AIS_Percentage_Error"] = np.where(
    result_df["Coal_Production"] != 0,
    (
        result_df["AIS_Absolute_Error"] /
        result_df["Coal_Production"].replace(0, np.nan)
    ) * 100,
    np.nan
)

result_df = result_df.reset_index(
    drop=True
)

result_df.to_csv(
    ROOT / "ais_result.csv",
    index=False
)

print("\nAIS Result CSV:")
print(result_df.head())

# ================================================================
# 20. AIS RESULT GRAPH
# ================================================================

plot_data = result_df.head(30)

positions = np.arange(
    len(plot_data)
)

plt.figure(figsize=(15, 7))

plt.plot(
    positions,
    plot_data["Coal_Production"],
    marker="o",
    label="Actual",
    linewidth=2
)

plt.plot(
    positions,
    plot_data["AIS_Prediction"],
    marker="s",
    label="AIS Prediction",
    linewidth=2
)

plt.plot(
    positions,
    plot_data["Baseline_Prediction"],
    linestyle="--",
    label="Baseline Prediction"
)

plt.title(
    "AIS Actual vs Predicted Coal Production",
    fontsize=15
)

plt.xlabel("Test Record")
plt.ylabel("Coal Production")

plt.legend()
plt.tight_layout()

plt.savefig(
    ROOT / "ais_result_graph.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ================================================================
# 21. AIS PREDICTION CSV
# ================================================================

# Estimates for all observed records.
# This is not a future time-series forecast.

all_ais_predictions = ais_model.predict(X)

all_baseline_predictions = baseline_model.predict(X)

all_nn_features = nn_preprocessor.transform(X)

all_ann_predictions = target_scaler.inverse_transform(
    ann.predict(
        all_nn_features,
        verbose=0
    )
).ravel()

prediction_df = data[
    [
        "Mine",
        "Company",
        "Coal_Production"
    ]
].copy()

prediction_df["AIS_Predicted_Production"] = (
    all_ais_predictions
)

prediction_df["Baseline_Predicted_Production"] = (
    all_baseline_predictions
)

prediction_df["ANN_Predicted_Production"] = (
    all_ann_predictions
)

prediction_df["Data_Split"] = np.where(
    prediction_df.index.isin(X_test.index),
    "Test",
    "Train"
)

prediction_df.to_csv(
    ROOT / "ais_prediction.csv",
    index=False
)

print("\nAIS Prediction CSV:")
print(prediction_df.head())

# ================================================================
# 22. AIS PREDICTION GRAPH
# ================================================================

plt.figure(figsize=(10, 8))

plt.scatter(
    y_test,
    ais_pred,
    alpha=0.75,
    s=65,
    label="AIS Optimized RF"
)

plt.scatter(
    y_test,
    baseline_pred,
    alpha=0.45,
    s=45,
    marker="x",
    label="Baseline RF"
)

minimum = min(
    float(y_test.min()),
    float(np.min(ais_pred)),
    float(np.min(baseline_pred))
)

maximum = max(
    float(y_test.max()),
    float(np.max(ais_pred)),
    float(np.max(baseline_pred))
)

plt.plot(
    [minimum, maximum],
    [minimum, maximum],
    "r--",
    label="Perfect Prediction"
)

plt.xlabel("Actual Coal Production")
plt.ylabel("Predicted Coal Production")

plt.title(
    "AIS Coal Production Prediction Graph",
    fontsize=15
)

plt.legend()
plt.tight_layout()

plt.savefig(
    ROOT / "ais_prediction_graph.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ================================================================
# 23. AIS CONVERGENCE GRAPH
# ================================================================

plt.figure(figsize=(11, 6))

plt.plot(
    history_df["Generation"],
    history_df["Best_CV_RMSE"],
    marker="o",
    linewidth=2,
    label="Best Cross-Validation RMSE"
)

plt.title(
    "AIS Optimization Convergence",
    fontsize=15
)

plt.xlabel("Generation")
plt.ylabel("Best CV RMSE (Lower Is Better)")

plt.legend()
plt.tight_layout()

plt.savefig(
    ROOT / "ais_convergence_graph.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ================================================================
# 24. ANN TRAINING GRAPH
# ================================================================

plt.figure(figsize=(11, 6))

plt.plot(
    ann_history.history["loss"],
    label="Training Loss"
)

plt.plot(
    ann_history.history["val_loss"],
    label="Validation Loss"
)

plt.title("AIS Project - ANN Training History")

plt.xlabel("Epoch")
plt.ylabel("Scaled MSE")

plt.legend()
plt.tight_layout()

plt.savefig(
    ROOT / "ais_ann_training_graph.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ================================================================
# 25. COMPANY PERFORMANCE GRAPH
# ================================================================

plt.figure(figsize=(13, 7))

sns.barplot(
    data=company_summary,
    x="Company",
    y="Total_Production",
    color="teal"
)

plt.title(
    "AIS Company-Wise Coal Production Analysis"
)

plt.xlabel("Company")
plt.ylabel("Total Coal Production")

plt.xticks(rotation=45, ha="right")
plt.tight_layout()

plt.savefig(
    ROOT / "ais_company_production_graph.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ================================================================
# 26. SAVE AIS CONFIGURATION
# ================================================================

config = {
    "project": (
        "AI-Based Coal Mine Production Analysis, "
        "Mining Company Performance Assessment, "
        "and Production Optimization System"
    ),
    "algorithm": "Artificial Immune System",
    "ais_variant": "Clonal Selection Algorithm",
    "dataset": CSV_PATH.name,
    "target": "Coal_Production",
    "features": ["Company"],
    "population_size": POPULATION_SIZE,
    "generations": GENERATIONS,
    "selected_count": SELECTED_COUNT,
    "random_replacements": RANDOM_REPLACEMENTS,
    "fitness_metric": "Cross-validation RMSE",
    "best_hyperparameters": best_antibody,
    "best_cv_rmse": float(best_fitness),
    "test_size": 0.20,
    "random_seed": SEED
}

with open(
    ROOT / "ais_config.yaml",
    "w",
    encoding="utf-8"
) as file:

    yaml.safe_dump(
        config,
        file,
        sort_keys=False
    )

# ================================================================
# 27. SAVE AIS JSON METADATA
# ================================================================

metadata = {
    "project": config["project"],
    "algorithm": "AIS - CLONALG",
    "dataset": CSV_PATH.name,
    "records": int(len(data)),
    "companies": int(data["Company"].nunique()),
    "mines": int(data["Mine"].nunique()),
    "best_hyperparameters": best_antibody,
    "best_cross_validation_rmse": float(best_fitness),
    "baseline_metrics": baseline_metrics,
    "ais_metrics": ais_metrics,
    "ann_metrics": ann_metrics,
    "evaluated_configurations": len(fitness_cache),
    "notes": (
        "AIS optimizes Random Forest hyperparameters. "
        "The H5 model is a separate ANN comparison model. "
        "Results are based on company-level features."
    )
}

with open(
    ROOT / "ais_model_metadata.json",
    "w",
    encoding="utf-8"
) as file:

    json.dump(
        metadata,
        file,
        indent=4
    )

# ================================================================
# 28. VERIFY GENERATED FILES
# ================================================================

required_files = [
    "ais_model.pkl",
    "ais_model.h5",
    "ais_config.yaml",
    "ais_model_metadata.json",
    "ais_accuracy_graph.png",
    "ais_heatmap.png",
    "ais_comparison_graph.png",
    "ais_result.csv",
    "ais_result_graph.png",
    "ais_prediction.csv",
    "ais_prediction_graph.png",
    "ais_convergence_graph.png",
    "ais_optimization_history.csv"
]

print("\n" + "=" * 70)
print("AIS OUTPUT VERIFICATION")
print("=" * 70)

for filename in required_files:

    path = ROOT / filename

    status = (
        "CREATED"
        if path.exists()
        else "MISSING"
    )

    print(f"[{status}] {filename}")

print("\n" + "=" * 70)
print("FINAL AIS PERFORMANCE")
print("=" * 70)

print(comparison_df.to_string(index=False))

print("\nBest AIS Parameters:")
print(best_antibody)

print("\nBest CV RMSE:", best_fitness)

print("\nAll AIS Files Saved In:")
print(ROOT)

print("\nAIS PROJECT EXECUTION COMPLETED!")
```
